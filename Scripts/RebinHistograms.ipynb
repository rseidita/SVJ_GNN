```python
import ROOT
import numpy as np
```

```python
def optimize_binning(histogram, n_min_mc_events_bin):

    #get the bin edges of the histogram
    bin_edges = []

    #take the binning from the signal region histogram
    for bin_i in range(1,histogram.GetNbinsX()+1):
        bin_edges.append(histogram.GetBinLowEdge(bin_i))
    bin_edges.append(histogram.GetBinLowEdge(histogram.GetNbinsX()+1))

    #define new_bin_edges list
    new_bin_edges = []

    #loop over the bins of the histogram and merge them if the number of events is less than n_min_mc_events_bin
    i = 0
    while i < len(bin_edges)-1:
        bin_i = i
        n_events = 0
        while n_events < n_min_mc_events_bin/117100 and bin_i < len(bin_edges)-1:
            n_events += histogram.GetBinContent(histogram.FindBin(bin_edges[bin_i]))
            bin_i += 1
        new_bin_edges.append(bin_edges[i])
        i = bin_i

    new_bin_edges.append(bin_edges[-1])

    #now rebin the histogram with the new binning
    new_bin_edges_array = np.array(new_bin_edges)

    return new_bin_edges_array


def apply_optimized_binning(histogram, new_bin_edges_array):
    
        #rebin the histogram
        histogram_rebinned = histogram.Rebin(len(new_bin_edges_array)-1, histogram.GetName(), new_bin_edges_array)
    
        return histogram_rebinned
```

```python
tfile = ROOT.TFile.Open("Bkg.root")
histogram = tfile.Get("Bkg_2SVJ")
n_min_mc_events_bin = 50

new_bin_edges_array = optimize_binning(histogram, n_min_mc_events_bin)
hist_rebined = apply_optimized_binning(histogram, new_bin_edges_array)

output_file = ROOT.TFile("Bkg_new.root", "RECREATE")
hist_rebined.Write(f"Bkg_2SVJ")
output_file.Close()
```

## Rebin 0,1 SVJ by factor and 2 SVJ by width

```python
import os
import ROOT
import numpy as np
ROOT.TH1.SetDefaultSumw2()
```

```python

wp = "0.50"
directory = f"/t3home/rhampp/SVJ_GNN/MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}"



# Define variable bin edges for 2SVJ
# variable_bins_2SVJ = [650, 800, 950, 1150, 1400, 2000]  # Example bin edges, adjust as needed
variable_bins_2SVJ = [650, 750, 850, 1000, 1150, 1300, 1450, 1600, 1750, 1900, 2000]
variable_bins_2SVJ = np.array(variable_bins_2SVJ, dtype = float)


for subdir in os.listdir(directory):
    for file in os.listdir(os.path.join(directory, subdir)):
        file_path = os.path.join(directory, subdir, file)
        file_hist = ROOT.TFile.Open(file_path)
        histogram = file_hist.Get(file.split(".")[0] + f"_{subdir}")
        
        # Determine binning method based on subdir
        if subdir == "0SVJ":
            nbin = 1
            rebinned_histogram = histogram.Rebin(nbin, file.split(".")[0])
        elif subdir == "1SVJ":
            nbin = 2
            rebinned_histogram = histogram.Rebin(nbin, file.split(".")[0])
        elif subdir == "2SVJ":
            # Use variable binning for 2SVJ
            rebinned_histogram = ROOT.TH1F(file.split(".")[0], histogram.GetTitle(), len(variable_bins_2SVJ) - 1, variable_bins_2SVJ)
            
            # Fill the new histogram with the data from the original one
            for i in range(1, histogram.GetNbinsX() + 1):
                content = histogram.GetBinContent(i)
                error = histogram.GetBinError(i)
                bin_center = histogram.GetBinCenter(i)
                
                # Find the appropriate bin for this data point
                new_bin = rebinned_histogram.FindBin(bin_center)
                
                # Add the content and error to the appropriate bin
                rebinned_histogram.SetBinError(new_bin, (rebinned_histogram.GetBinError(new_bin)**2 + error**2)**0.5)
                rebinned_histogram.SetBinContent(new_bin, rebinned_histogram.GetBinContent(new_bin) + content)

        # Create output directory structure if it doesn't exist
        rebinned_dir = directory + "_rebinned"
        if not os.path.exists(rebinned_dir):
            os.mkdir(rebinned_dir)
        subdir_path = os.path.join(rebinned_dir, subdir)
        if not os.path.exists(subdir_path):
            os.mkdir(subdir_path)

        # Write the rebinned histogram to a new file
        output_path = os.path.join(subdir_path, file)
        output_file = ROOT.TFile(output_path, "RECREATE")
        rebinned_histogram.Write()
        output_file.Close()

        # Close the input file
        file_hist.Close()
```
