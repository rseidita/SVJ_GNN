```python
import ROOT
```

```python
hist_output_directory = "/t3home/ceholm/SVJ_GNN/tagged_jets_hist/ROOT/12_08/new/"


# Open the ROOT file
file_hist = ROOT.TFile(f"{hist_output_directory}/qcd_epoch_400_bs_256_lr_0.007_flt_0.55_arch_100_to_10_with_4momenta_large_lr_scheduling_kt_cut_weights_fix_GCNConv_mdmm_constrain_node_loss_1jetshist.root", "READ")

# Retrieve the histogram
hist = file_hist.Get(f"qcd_1jets")

# Ensure the histogram has sum of squares of weights enabled
hist.Sumw2()
```

```python

# Prepare to extract the weights of each event
weights = []

# Loop over all bins and extract the weights
for bin_idx in range(1, hist.GetNbinsX() + 1):
    bin_content = hist.GetBinContent(bin_idx)
    bin_error = hist.GetBinError(bin_idx)
    for _ in range(int(bin_content)):
        # Assuming you want to retrieve the bin weights
        weights.append(bin_content)

# Close the ROOT file
file_hist.Close()

# Now 'weights' contains the list of weights for the events
```
