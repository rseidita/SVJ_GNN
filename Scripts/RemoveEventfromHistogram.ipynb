```python
import ROOT
```

```python

# Open the ROOT file
file = ROOT.TFile.Open("Wjets.root", "UPDATE")

# Get the histogram
hist = file.Get("Wjets_0SVJ")
```

```python

# Define the bin ranges you want to remove
bin_to_remove_1 = hist.FindBin(1460)
# bin_to_remove_2 = hist.FindBin(1825)
```

```python
bin_to_remove_1
```

```python
hist.GetBinContent(56)
```

```python

# Set the bin content and error to zero for the selected bins
hist.SetBinContent(bin_to_remove_1, 0)
hist.SetBinError(bin_to_remove_1, 0)

# hist.SetBinContent(bin_to_remove_2, 0)
# hist.SetBinError(bin_to_remove_2, 0)


```

```python


# Save the changes to the ROOT file
hist.Write(hist.GetName(), ROOT.TObject.kOverwrite)
file.Close()

print("Bins around 1100 and 1800 removed.")
```
