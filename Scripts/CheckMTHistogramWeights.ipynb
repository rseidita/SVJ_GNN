```python
import ROOT
```

```python

# Define the path to the ROOT file
name = "Bkg"
cat = "1SVJ"
root_file_path = "/t3home/rhampp/SVJ_GNN/MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-0.60_newbinning2/"+cat+ "/"+name +".root"

# Open the ROOT file
file = ROOT.TFile(root_file_path, "READ")

# Check if the file is successfully opened
if file.IsOpen():
    print("File successfully opened.")
else:
    print("Error opening file.")
    exit()

# Assuming the histogram name is known (e.g., "hist_name")
hist_name = name +"_" + cat  # Replace with the actual histogram name

# Retrieve the histogram from the file
hist = file.Get(hist_name)

# Check if the histogram is successfully retrieved
if not hist:
    print(f"Error: Histogram '{hist_name}' not found in the file.")
    exit()

# Loop over the bins of the histogram and print the bin contents (weights)
for bin in range(1, hist.GetNbinsX() + 1):
    content = hist.GetBinContent(bin)
    error = hist.GetBinError(bin)

    print(f"Bin {bin}: {content}, {error}")

# Close the ROOT file
file.Close()
```
