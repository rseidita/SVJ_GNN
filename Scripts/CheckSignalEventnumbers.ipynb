```python
import os
import glob
import uproot
import numpy as np

# Define the directories to search
dirs = [
    "/pnfs/psi.ch/cms/trivcat/store/t3groups/ethz-susy/darkshowers/samples/signal_scouting/PFNano/SVJ_std_UL2018_scouting_test_new/",
    "/pnfs/psi.ch/cms/trivcat/store/t3groups/ethz-susy/darkshowers/samples/signal_scouting/PFNano/for_trigger_study/SVJ_std_UL2018_scouting_test_large/",
    "/pnfs/psi.ch/cms/trivcat/store/t3groups/ethz-susy/darkshowers/samples/signal_scouting/PFNano/for_trigger_study/SVJ_std2_UL2018_scouting_off_puppi_NO_HLT_JECS/",
    "/pnfs/psi.ch/cms/trivcat/store/t3groups/darkshowers/samples_scouting/NANOAOD/"
]

# Define the signal keys
signal_keys = [
    "mMed-700GeV_mDark-20GeV_rinv-0.3_alpha-peak",
    "mMed-700GeV_mDark-20GeV_rinv-0.5_alpha-peak",
    "mMed-700GeV_mDark-20GeV_rinv-0.7_alpha-peak",
]

# Initialize the Signal dictionary
Signal = {}
for key in signal_keys:
    Signal[key] = []

# Function to find subdirectories containing the signal key
def find_signal_dirs(base_dir, signal_key):
    signal_dirs = []
    for root, dirs, files in os.walk(base_dir):
        for dir_name in dirs:
            if signal_key in dir_name:
                signal_dirs.append(os.path.join(root, dir_name))
    return signal_dirs

# Loop over each signal key
for key in signal_keys:
    print(f"Processing key: {key}")
    for base_dir in dirs:
        signal_dirs = find_signal_dirs(base_dir, key)
        
        for signal_dir in signal_dirs:
            # Find all ROOT files in the current signal directory
            file_paths = glob.glob(os.path.join(signal_dir, "*.root"))
            
            # Check if any files are found
            if not file_paths:
                print(f"No ROOT files found in directory: {signal_dir}")
                continue
            
            # Concatenate the "event" branch from all files in the current directory
            try:
                events = uproot.concatenate([f"{path}:mmtree/Events" for path in file_paths], filter_name="event")
                Signal[key].append(events["event"].to_numpy())
            except Exception as e:
                print(f"Failed to concatenate branches in {signal_dir}: {e}")

    # Concatenate all events for the current key
    if Signal[key]:
        Signal[key] = np.concatenate(Signal[key])
    else:
        Signal[key] = np.array([])

print("Signal:", Signal)
```

```python
from collections import Counter

count = {}

for key in Signal:
    count[key] = Counter(Signal[key])
```
