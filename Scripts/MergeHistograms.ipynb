```python
import ROOT
ROOT.TH1.SetDefaultSumw2()
ROOT.TH1.StatOverflows(True)

```

```python
cats = [ "0SVJ", "1SVJ", "2SVJ"]
wp = "0.50"

for cat in cats:    
    # Open the ROOT files
    file1 = ROOT.TFile.Open(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/QCD.root")
    file2 = ROOT.TFile.Open(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/TTbar.root")
    file3 = ROOT.TFile.Open(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/Wjets.root")
    file4 = ROOT.TFile.Open(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/Zjets.root")

    # Retrieve histograms from the files
    hist1 = file1.Get(f"QCD_{cat}")
    hist2 = file2.Get(f"TTbar_{cat}")
    hist3 = file3.Get(f"Wjets_{cat}")
    hist4 = file4.Get(f"Zjets_{cat}")
    print("done1")

    # Check that histograms are correctly retrieved
    if not hist1 or not hist2 or not hist3 or not hist4:
        print("Error: One or more histograms could not be found.")
        exit()

    # Merge histograms
    hist1.Add(hist2)
    hist1.Add(hist3)
    hist1.Add(hist4)

    print("done2")
    # Save the merged histogram into a new file
    output_file = ROOT.TFile(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/Bkg.root", "RECREATE")
    hist1.Write(f"Bkg_{cat}")


    output_file.Close()

    # Close the original files
    file1.Close()
    file2.Close()
    file3.Close()
    file4.Close()
    print(f"MtDistributions/LundNet5+_SVJ-QCD-TTJet-EW_vs12/wp-{wp}/{cat}/Bkg.root", "processed")
```
