```python
import os
import shutil
import re
```

```python

# Define the model names
model_names = [
    # "QCD_HT1000to1500",
    # "QCD_HT1500to2000",
    "QCD_HT500to700",
    # "QCD_HT700to1000",
    # "QCD_HT2000toInf",
    "QCD_HT300to500",
]

# Define the directory containing the files
source_dir = "/pnfs/psi.ch/cms/trivcat/store/user/rhampp/forTraining/Preselection_applied/Bkg/QCD_new/"  # Update with your source directory path

# output_dir = "/pnfs/psi.ch/cms/trivcat/store/user/rhampp/forTraining/Preselection_applied/Bkg_sorted/QCD/"

# Function to extract the numeric part from the filename for sorting
def extract_number(filename):
    match = re.search(r'part-(\d+)', filename)
    if match:
        return int(match.group(1))
    return -1

# Create directories if they don't exist
for model_name in model_names:
    os.makedirs(os.path.join(source_dir, model_name), exist_ok=True)

# Get the list of files in the source directory
files = os.listdir(source_dir)

# Dictionary to hold lists of files for each model name
sorted_files = {model_name: [] for model_name in model_names}

# Categorize files into respective model name lists
for file in files:
    for model_name in model_names:
        if model_name in file:
            sorted_files[model_name].append(file)
            break

# Sort files within each model name list and move them to the respective directories
for model_name, file_list in sorted_files.items():
    file_list.sort(key=extract_number)
    for file in file_list:
        if ".root" in file:
            src_path = os.path.join(source_dir, file)
            dest_path = os.path.join(source_dir, model_name, file)
            shutil.copy(src_path, dest_path)

print("Files have been sorted and copied to their respective directories.")
```
