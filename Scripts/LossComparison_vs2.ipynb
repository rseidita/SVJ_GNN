```python
import os
import uproot
import matplotlib.pyplot as plt
```

### Comparison between different models

```python

## Comparison between different Models
path = os.getcwd()
path = os.path.join(path, "output")
model = "_SVJ-QCD_"
data = {
    "LundNet3": None,
    "LundNet3+": None,
    "LundNet5": None,
    "LundNet5+": None,
}
versions = ["vs1", "vs1", "vs1", "vs2"] 
data = {
    "LundNet3+": None,
    "LundNet5": None,
    "LundNet5+": None,
}
versions = ["vs1", "vs1", "vs2"] # corresponding version for lr = 0.001

lr = [0.001] # corresponding learning rate

for key, version in zip(data,versions):
    filepath = os.path.join(path, key + model + version)
    data[key]= uproot.open(filepath + "/" + key + model + version + "_loss.root:Output").arrays()
colors = [ '#E63946', '#F4A261',  '#A8DADC','#457B9D',]


for n,key in enumerate(data):
    plt.plot(data[key]["index"], data[key]["Training_loss"], color = colors[n], linewidth = 0.9, linestyle = '-', label = key)
    plt.plot(data[key]["index"], data[key]["Validation_loss"], color = colors[n],linewidth = 0.9,  linestyle = ':')

plt.yscale('log')
plt.legend()
first_legend = plt.legend(loc=[0.72,0.8])
ax = plt.gca().add_artist(first_legend)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.ylim(0.49,0.8)
legend_handles = [plt.Line2D([0], [1], color='black',linewidth = 0.9, linestyle=linestyle, label=label) for label, linestyle in zip(["Training", "Validation"],['-',':'])]


plt.legend(handles=legend_handles, loc=[0.47,0.86])


plt.plot(figsize=(10, 5))
plot_name = "LundNet_LossComparison.png"
plt.savefig(os.path.join(path,plot_name), dpi=1200)
plt.show()
```

### Comparison of different patience

```python
path = os.getcwd()
path = os.path.join(path, "output")
model = "LundNet5+_SVJ-QCD_"

data = {
    "vs4": None,
    "vs2": None,
    "vs3" : None,
    "vs1":None,
}
label = ["Patience = 100","Patience = 50", "Patience = 30", "Patience = 10"] # corresponding learning rate

for key in data:
    filepath = os.path.join(path, model + key)
    data[key]= uproot.open(filepath + "/" + model + key + "_loss.root:Output").arrays()


colors = ['#457B9D',   '#F4A261','#E63946', 'plum']


for n,key in enumerate(data):
    plt.plot(data[key]["index"], data[key]["Training_loss"], color = colors[n], linewidth = 0.9, linestyle = '-', label = label[n])
    plt.plot(data[key]["index"], data[key]["Validation_loss"], color = colors[n],linewidth = 0.9,  linestyle = ':')

plt.yscale('log')
plt.legend()
first_legend = plt.legend(loc=[0.67,0.75])
ax = plt.gca().add_artist(first_legend)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.ylim(0.49,0.8)
legend_handles = [plt.Line2D([0], [1], color='black',linewidth = 0.9, linestyle=linestyle, label=label) for label, linestyle in zip(["Training", "Validation"],['-',':'])]


plt.legend(handles=legend_handles, loc=[0.42,0.86])
plt.title("LundNet5+")

plt.plot(figsize=(10, 5))
plot_name = "LundNet5+_LossComparisonPatience.png"
plt.savefig(os.path.join(path,plot_name), dpi=1200)
plt.show()
```
