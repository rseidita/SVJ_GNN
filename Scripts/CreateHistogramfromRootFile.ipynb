```python

# Define the base directory
base_directory_bkg = "/pnfs/psi.ch/cms/trivcat/store/user/rhampp/forTraining/Preselection_applied/Bkg_sorted/"

# List of datasets with corresponding subdirectories
all_datasets = [
    ["QCD/QCD_HT300to500",
     "QCD/QCD_HT500to700",
     "QCD/QCD_HT700to1000",
     "QCD/QCD_HT1000to1500", 
     "QCD/QCD_HT1500to2000", 
     "QCD/QCD_HT2000toInf",],
    # ["/TTJet"],
    # ["EW/WJets_inclusive"],
    # ["EW/ZJetsToNuNu_HT-100To200",
    #  "EW/ZJetsToNuNu_HT-200To400",
    #  "EW/ZJetsToNuNu_HT-400To600",
    #  "EW/ZJetsToNuNu_HT-600To800",
    #  "EW/ZJetsToNuNu_HT-800To1200",
    #  "EW/ZJetsToNuNu_HT-1200To2500",
    #  "EW/ZJetsToNuNu_HT-2500ToInf"],
]

names = ["QCD", "TTbar", "Wjets", "Zjets"]
names = [ "QCD"]

# Loop over each dataset group
for dataset_group, name in zip(all_datasets, names):
    # Combine the output file name from the dataset group (for example purposes, join the last directory name)
    output_file_name = name

    # Define a histogram for MT (you may define different histograms for different datasets)
    hist_MT = ROOT.TH1F(output_file_name, f"Histogram of MT for {output_file_name}", 100, 650, 2000)

    # Loop over each dataset within the group
    for dataset in dataset_group:
        # List all files in the directory
        all_files = os.listdir(base_directory_bkg+ dataset)
        
        # Filter the ROOT files (you can adjust the filter if needed)
        root_files = [f for f in all_files if f.endswith(".root")]
        weight = fetch_weight(dataset.split("/")[1])
        # Process each ROOT file
        for root_file in root_files:
            input_file_path = os.path.join(base_directory_bkg+ dataset, root_file)

            # Open the original ROOT file
            input_file = ROOT.TFile(input_file_path, "READ")
        
            try:
                tree = input_file.Get("Events")
                i = 0
                # Loop over the entries in the tree and fill the histogram
                for entry in tree:
                    hist_MT.Fill(entry.MT01FatJetMET, weight)
                
            except:
                continue
            # Close the input file
            print("processed:", input_file )

            input_file.Close()
    
    # Create a new ROOT file to save the combined histogram for this dataset group
    output_file = ROOT.TFile(f"{output_file_name}.root", "RECREATE")
    
    # Write the histogram to the new file
    hist_MT.Write()
    
    # Close the output file
    output_file.Close()

    print(f"Histogram of MT saved to {output_file_name}.root")
    

print("All histograms have been saved.")
```
