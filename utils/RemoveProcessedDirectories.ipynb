```python
import os
import shutil
```

```python

def remove_proceses_directories(root_directory):
    # Traverse the directory tree
    for dirpath, dirnames, filenames in os.walk(root_directory):
        # Iterate over a copy of dirnames to avoid modifying the list while iterating
        for dirname in dirnames[:]:
            if dirname == "processed":
                full_dir_path = os.path.join(dirpath, dirname)
                print(f"Removing directory: {full_dir_path}")
                shutil.rmtree(full_dir_path)
                # Optionally, remove dirname from dirnames to prevent further processing
                dirnames.remove(dirname)
```

```python


import os

def remove_pre_files(root_directory):
    # Traverse the directory tree
    for dirpath, dirnames, filenames in os.walk(root_directory):
        # Check if the file "pre_filter.pt" is in the current directory
        if 'pre_filter.pt' in filenames:
            full_file_path = os.path.join(dirpath, 'pre_filter.pt')
            print(f"Removing file: {full_file_path}")
            os.remove(full_file_path)
        if 'pre_transform.pt' in filenames:
            full_file_path = os.path.join(dirpath, 'pre_transform.pt')
            print(f"Removing file: {full_file_path}")
            os.remove(full_file_path)
```

```python
import fnmatch

def remove_graph_files(root_directory, pattern):
    # Pattern to match files like SVJ_<variable1>_<variable2>_LN3.pt
    
    # Traverse the directory tree
    for dirpath, dirnames, filenames in os.walk(root_directory):
        for filename in filenames:
            # Check if the filename matches the pattern
            if fnmatch.fnmatch(filename, pattern):
                full_file_path = os.path.join(dirpath, filename)
                print(f"Removing file: {full_file_path}")
                os.remove(full_file_path)
```
